```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
retval = cv2.imread("WhatsApp Image 2026-11-23 at 12.53.49.jpeg", cv2.IMREAD_COLOR)

print(retval.shape)
print(retval.dtype)

#winston = retval[:, :, ::-1]


b, g, r = cv2.split(retval)

# the 142 is such that each individual value represents rows, coloumns and index.
plt.subplot(141); plt.imshow(r, cmap='gray'); plt.title("red channel")
plt.subplot(142); plt.imshow(g, cmap='gray'); plt.title("green channel")
plt.subplot(143); plt.imshow(b, cmap='gray'); plt.title("blue channel")

merged = cv2.merge((b, g, r))

plt.subplot(144); plt.imshow(merged[:, :, ::-1]); plt.title("Merged Output")
# plt.imshow(retval, cmap='gray')
```
